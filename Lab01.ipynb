```python
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

figure = plt.figure(1, figsize = (12, 4))
subplot3d = plt.subplot(111, projection='3d')
surface = subplot3d.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap=matplotlib.cm.coolwarm, linewidth=0.1)
plt.show()
```

```python
import pandas as pd
```

```python
!wget -q https://elitedatascience.com/wp-content/uploads/2022/07/Pokemon.csv
# Read dataset
df = pd.read_csv('Pokemon.csv', index_col=0, encoding='latin')
```

```python
features = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
```

```python
features = ['HP', 'Attack', 'Defense', 'Sp. Ark', 'Sp. Def', 'Speed']
sub_df = df[features]
```

```python
pearson_corr = sub_df.corr(method='pearson')
spearman_corr = sub_df.corr(method='spearman')

# 2. Visualize Pearson Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(pearson_corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Pearson Standard Coefficient Heatmap')
plt.show()

# 3. Visualize Spearman Rank Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    spearman_corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1
)
plt.title('Spearman Rank Correlation Heatmap')
plt.show()
```
